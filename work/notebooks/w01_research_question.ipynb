## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Path relative to work/notebooks/
try:
    df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
    print("Dataset loaded successfully!")
except FileNotFoundError:
    # Fallback path if running from the root directory
    df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
    print("Dataset loaded successfully via fallback path!")

# Calculate numbers
total_pages = df['content_id'].nunique()
declining_pages = df[df['trend_direction'] == 'down'].shape[0]
pct_declining = (declining_pages / len(df)) * 100

high_impact_decay = df[
    (df['trend_direction'] == 'down') & 
    (df['content_age_days'] >= 180) & 
    (df['impressions_90d'] >= 500)
].shape[0]

print(f"Total Pages: {total_pages}")
print(f"Declining Pages: {declining_pages} ({pct_declining:.1f}%)")
print(f"High-Impact Decay Candidates: {high_impact_decay}")
```
